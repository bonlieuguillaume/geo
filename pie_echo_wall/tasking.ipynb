# SAR passes over an AOI

When a SAR satellite can image an area, and at which incidence, from its orbit: a TLE, a NORAD id (latest TLE from CelesTrak), mean elements for a satellite not flying yet, or the operator's ephemerides (CCSDS OEM, Sentinel-1 `.EOF`, CSV). All the logic lives in `tasking.py`; this notebook only drives it — see `README.md`.

| Cell | What it does |
| --- | --- |
| 1 | Imports the module |
| 2 | The webmap: draw or paste the AOI, set orbit, sensor and window, *Find passes*, click a pass to see its swath |
| 3 | The last result as a DataFrame (`ui.passes`) |
| 4 | The same from plain Python, without the interface |

```python
import tasking
```

```python
ui = tasking.build_ui(center=(46.5, 2.5), zoom=6, days=12)
ui
```

```python
ui.passes
```

```python
sat = tasking.satellite(elements=dict(epoch="2026-10-05", repeat_cycle=(12, 175), ltan_hours=18.0))
# sat = tasking.satellite(norad_id=39634)
# sat = tasking.satellite(tle="""1 ...\n2 ...""")
# sat = tasking.satellite(ephemeris=r"C:\path\to\predicted_orbit.oem")  # OEM, S1 .EOF or CSV
print(tasking.orbit_summary(sat))

passes = tasking.find_passes(
    sat, "POLYGON((2.2 48.7, 2.6 48.7, 2.6 49.0, 2.2 49.0, 2.2 48.7))",
    "2026-10-05", "2026-10-17",
    look_side="right", incidence_range=(29.1, 46.0),
)
passes
```
